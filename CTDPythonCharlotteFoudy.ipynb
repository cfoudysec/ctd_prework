```python
# This script asks the user for the name and age in order to determine whether they are eligible to enter an all-expenses paid trip to Hawaii
```

```python
name = input ("What is your name? ")
age = int (input("How old are you? "))
```

```python
print(f"{name}, you are {age} years old.")
```

```python
if age >= 18:
  print("You are eligible to enter the all-expenses-paid trip to Hawaii giveaway!")
else:
  print("You must be at least 18 years old to enter the Hawaii trip giveaway.")
```

```python
# Example Run 1:
# What is your name? Charlotte
# How old are you? 44
# Charlotte, you are 44 years old.
# You are eligible to enter the all-expenses-paid trip to Hawaii giveaway!

# Example Run 2:
# What is your name? Michael
# How old are you? 11
# Michael, you are 11 years old.
# You are not eligible to enter the all-expenses-paid trip to Hawaii giveway!
```
